```python
import requests
from bs4 import BeautifulSoup

class BookScraper:
    def __init__(self):
        self.base_url = 'https://www.valdemarsro.dk'
        self.start_url = f'{self.base_url}/opskrifter/'

    def scrape_all_recipes(self):
        all_recipes = []
        page = 1

        while True:
            url = f'{self.start_url}page/{page}/'
            print(f'Scraping page {page}: {url}')
            response = requests.get(url)
            if response.status_code != 200:
                print(f"Stopping at page {page}, got status code: {response.status_code}")
                break

            soup = BeautifulSoup(response.content, 'html.parser')
            recipes_on_page = self.scrape_recipes_from_soup(soup)

            if not recipes_on_page:
                print(f"No recipes found on page {page}, stopping.")
                break

            all_recipes.extend(recipes_on_page)
            page += 1

        return all_recipes

    def scrape_recipes_from_soup(self, soup):
        opskrift_details = []

        for opskrift in soup.find_all('div', class_='post-list-item'):
            title_tag = opskrift.find('span', class_='post-list-item-title')
            title = title_tag.get_text(strip=True) if title_tag else None
            
            a_tag = opskrift.find('a')
            
            img = opskrift.find('img')
            image_url = img.get('src') if img else None

            href = a_tag.get('href') if a_tag and a_tag.has_attr('href') else None
            product_page_url = href if href and href.startswith('http') else f'{self.base_url}{href}' if href else None

            opskrift_details.append({
                'Title': title,
                'Image URL': image_url,
                'Product Page URL': product_page_url
            })

        return opskrift_details
```

```python
scraper = BookScraper()
all_recipes = scraper.scrape_all_recipes()

print(f"Total recipes found: {len(all_recipes)}")
for recipe in all_recipes[0:5]:
    print(recipe)
```

```python
with open('opskrifer_valdemarsro.csv', 'w', newline='') as csvfile:
    fieldnames = ['Title', 'Image URL', 'Product Page URL']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(all_recipes)
```
